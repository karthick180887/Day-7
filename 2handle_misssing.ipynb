```python
print("Missing values in Percentage TRAIN (before)")
print(X_train.isna().mean()*100) #count of missing cells per column
print("_"*50)
print("Missing Values in Percentage TEST (before)")
print(X_test.isna().mean()*100)
```

```python
missing_threshold = 40

missing_percent_train = X_train.isna().mean() * 100
print("Percentage of missing in each column:")
print(missing_percent_train)

high_missing_cols = missing_percent_train[missing_percent_train > missing_threshold].index.tolist()

print("colunms with more then 40% missing values:")
print(high_missing_cols)
```

```python
X_train = X_train.drop(columns=high_missing_cols)
X_test = X_test.drop(columns=high_missing_cols)
```

```python
#Identify numerical and categorical Columns
```

```python
#Numerical(Continuous/Counts) - Mean or Median
#categorical Values = Mode(Most Frequent) or "Unkonwn"
```

```python
cols_list = X_train.columns.to_list()
print(cols_list)
```

```python
nums_cols = ["age","sibsp","parch","fare"]
cat_cols = ["pclass","gender","embarked","class","who","adult_male","embark_town","alone"]
```

```python
print("Numeric Columns:",nums_cols)
print("Caegorical Columns:", cat_cols)
```

```python
#impute the missing values only in Train only
```

```python
X_train_imputed = X_train.copy()
X_test_imputed = X_test.copy()
```

```python
#Numeric imputation
#Mean = Sum/Count, it is pulled towards extreme values
#Median = midlle values after sorting, its is robust to outliers and skewed data
```

```python
from sklearn.impute import SimpleImputer
num_imputer = SimpleImputer(strategy="median")
X_train[nums_cols] = num_imputer.fit_transform(X_train[nums_cols])
X_test[nums_cols] = num_imputer.transform(X_test[nums_cols])
```

```python
numeric_medians = {}
for col in nums_cols:
    median_val = X_train_imputed[col].median()
    numeric_medians[col] = median_val
    print(f"Filling Numeric Column {col} with Train Median: {median_val}")
    X_train_imputed[col] = X_train_imputed[col].fillna(median_val)
    X_test_imputed [col] = X_test_imputed[col].fillna(median_val)
```

```python
X_train_imputed
```

```python
X_test_imputed
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent") #strategy="constant" fill_value="Unknown"
X_train[cat_cols] = cat_imputer.fit_transform(X_train[cat_cols])
X_test[cat_cols] = cat_imputer.fit_transform(X_test[cat_cols])
```

```python
categorical_modes = {}
for col in cat_cols:
    mode_val = X_train_imputed[col].mode().iloc[0]
    categorical_modes[col] = mode_val
    print(f"Filling Categorical Column {col} with Train Mode: {mode_val}")
    X_train_imputed[col] = X_train_imputed[col].fillna(mode_val)
    X_test_imputed [col] = X_test_imputed[col].fillna(mode_val)
```

```python
print(X_train.isna().mean() * 100 )
print(X_test.isna().mean() * 100)
```

```python
train_imputed_df = X_train_imputed.copy()
test_imputed_df = X_test_imputed.copy()

train_imputed_df[target] = y_train.values
test_imputed_df[target] = y_test.values

train_imputed_df.to_csv("titanic_train_imputed.csv", index=False)
test_imputed_df.to_csv("titanic_test_imputed.csv", index=False)
print("\n Saved Imputed Data")
```
